Post-covid discrepancy will be issue if models are blindly trained on pre-covid data

### Adding Covid flag to features

```python
# Add COVID regime indicator
# df['is_post_covid'] = (df['Year'] >= 2023).astype(int)

# print("Pre-COVID rows:", (df['is_post_covid'] == 0).sum())
# print("Post-COVID rows:", (df['is_post_covid'] == 1).sum())
# print(df.groupby('is_post_covid')['weekly_exits'].mean().round(0))
```

```python
features = [
    'Month',
    'Week',
    'population_density',
    'low_car_rate',
    'transit_commuter_rate',
    'driver_rate',
    'carpool_rate',
    'employment_rate',
    'renter_rate',
    'median_income',
    'avg_trip_distance_km',
    'num_unique_destinations',
    'is_weekend_ratio',
    'is_post_covid'  # new feature
]

# Redo splits with new feature
train = df[df['Year'].isin([2017, 2018, 2019, 2023])]
val = df[df['Year'] == 2024]
test = df[df['Year'].isin([2025, 2026])]

X_train = train[features]
y_train = train[target]
X_val = val[features]
y_val = val[target]
X_test = test[features]
y_test = test[target]

# Retrain linear regression
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_val)

rmse_lr = np.sqrt(mean_squared_error(y_val, y_pred_lr))
mae_lr = mean_absolute_error(y_val, y_pred_lr)

print("\nLinear Regression Results with COVID indicator:")
print(f"RMSE: {rmse_lr:,.0f}")
print(f"MAE: {mae_lr:,.0f}")
print(f"MAE as % of mean: {(mae_lr / y_val.mean() * 100):.1f}%")

plt.figure(figsize=(8, 6))
plt.scatter(y_val, y_pred_lr, alpha=0.3, color='steelblue')
plt.plot([0, y_val.max()], [0, y_val.max()], 'r--', label='Perfect prediction')
plt.xlabel('Actual Weekly Exits')
plt.ylabel('Predicted Weekly Exits')
plt.title('Linear Regression: Actual vs Predicted')
plt.legend()
plt.tight_layout()
plt.savefig('../../notebooks/figures/lr_actual_vs_predicted.png', dpi=150)
plt.show()
```

### XGBoost Model

```python

# Train XGBoost
xgb_model = xgb.XGBRegressor(
    n_estimators = 500,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.9,
    colsample_bytree=0.9,
    random_state=42,
    early_stopping_rounds=20
)

xgb_model.fit(
    X_train, y_train,
    eval_set=[(X_train, y_train), (X_val, y_val)],
    verbose=50
)

y_pred_xgb = xgb_model.predict(X_val)

rmse_xgb = np.sqrt(mean_squared_error(y_val, y_pred_xgb))
mae_xgb = mean_absolute_error(y_val, y_pred_xgb)

print("\nXGBoost Results:")
print(f"RMSE: {rmse_xgb:,.0f}")
print(f"MAE: {mae_xgb:,.0f}")
print(f"MAE as % of mean: {(mae_xgb / y_val.mean() * 100):.1f}%")

results = xgb_model.evals_result()

plt.figure(figsize=(10, 5))
plt.plot(results['validation_0']['rmse'], label='Train RMSE', color='steelblue')
plt.plot(results['validation_1']['rmse'], label='Validation RMSE', color='coral')
plt.xlabel('Number of Trees')
plt.ylabel('RMSE')
plt.title('XGBoost Training vs Validation RMSE')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../../notebooks/figures/xgb_training_curve.png', dpi=150)
plt.show()
```
