```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.model_selection import cross_val_score, cross_val_predict
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
import xgboost as xgb
import matplotlib.pyplot as plt
MODEL_PATH = '../../models/xgb_model_2.json'
```

```python
xgb_model = xgb.XGBRegressor()
xgb_model.load_model(MODEL_PATH)
df_model = pd.read_csv('../../data/processed/df_model.csv')
```

```python
SITE_FEATURES = [
    'Month',
    'Week',
    'population_density',
    'low_car_rate',
    'transit_commuter_rate',
    'driver_rate',
    'carpool_rate',
    'employment_rate',
    'renter_rate',
    'median_income',
    'is_post_covid',
    'dist_to_nearest_job_center_km'
]
```

```python
def dist_to_point(lat, lon, clat, clon):
    R = 6371
    dlat, dlon = np.radians(lat - clat), np.radians(lon - clon)
    a = np.sin(dlat/2)**2 + np.cos(np.radians(lat))*np.cos(np.radians(clat))*np.sin(dlon/2)**2
    return 2 * R * np.arcsin(np.sqrt(a))

def dist_to_nearest_job_center(lat, lon, centers):
    dists = [dist_to_point(lat, lon, c[0], c[1]) for c in centers]
    return min(dists)
```

```python
import requests
import geopandas as gpd
import numpy as np

# Load TIGER tract boundaries
tract_url = 'https://www2.census.gov/geo/tiger/TIGER2023/TRACT/tl_2023_06_tract.zip'
ca_tract_geo = gpd.read_file(tract_url)

BAY_AREA_COUNTIES = ['001','013','041','055','075','081','085','095','097']
ca_tract_geo = ca_tract_geo[ca_tract_geo['COUNTYFP'].isin(BAY_AREA_COUNTIES)]
ca_tract_geo = ca_tract_geo.to_crs('EPSG:4326')

# Load LODES job data
lodes_url = 'https://lehd.ces.census.gov/data/lodes/LODES8/ca/wac/ca_wac_S000_JT00_2021.csv.gz'
lodes = pd.read_csv(lodes_url, compression='gzip', dtype={'w_geocode': str})
lodes['GEOID'] = lodes['w_geocode'].str[:11]

target_geoids = set(ca_tract_geo['GEOID'].values)
lodes = lodes[lodes['GEOID'].isin(target_geoids)]

jobs_by_tract = lodes.groupby('GEOID')['C000'].sum().reset_index()
jobs_by_tract.rename(columns={'C000': 'total_jobs'}, inplace=True)

# Merge onto tract geometry
tracts = ca_tract_geo[['GEOID', 'ALAND', 'geometry']].merge(
    jobs_by_tract, on='GEOID', how='left'
)
tracts['job_density'] = tracts['total_jobs'] / (tracts['ALAND'] / 1_000_000)

# Compute job centers
emp_threshold = np.percentile(tracts['job_density'].dropna(), 95)
job_center_tracts = tracts[tracts['job_density'] >= emp_threshold]

MIN_DIST_KM = 4
filtered_centers = []

for _, row in job_center_tracts.sort_values('job_density', ascending=False).iterrows():
    lat = row.geometry.centroid.y
    lon = row.geometry.centroid.x
    if not filtered_centers:
        filtered_centers.append((lat, lon))
        continue
    dists = [dist_to_point(lat, lon, c[0], c[1]) for c in filtered_centers]
    if min(dists) >= MIN_DIST_KM:
        filtered_centers.append((lat, lon))

job_centers = np.array(filtered_centers)
print(f'Job centers: {len(job_centers)}')

# Add to df_model
df_model['dist_to_nearest_job_center_km'] = df_model.apply(
    lambda r: dist_to_nearest_job_center(r['stop_lat'], r['stop_lon'], job_centers),
    axis=1
)
```

```python
test = df_model[df_model['Year'].isin([2025, 2026])]
```

```python
X_test = test[SITE_FEATURES]
y_test = test['weekly_exits']
```

```python
y_pred_log = xgb_model.predict(X_test)
y_pred = np.expm1(y_pred_log)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print(f'MAE:  {mae:.0f} ({mae/y_test.mean()*100:.1f}% of mean)')
print(f'RMSE: {rmse:.0f} ({rmse/y_test.mean()*100:.1f}% of mean)')
```
