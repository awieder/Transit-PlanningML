```python
import pandas as pd

# Load stops directly from the txt file
stops = pd.read_csv('../data/raw/gtfs/google_transit_20260112-20260807_v04/stops.txt')
print(stops.columns.tolist())
print(stops.head())
```

```python
# Keep only parent stations (one row per station)
stations = stops[stops['location_type'] == 1][['stop_name', 'stop_lat', 'stop_lon', 'zone_id']].copy()

# If that gives empty result, use zone_id to deduplicate instead
if len(stations) == 0:
    stations = stops.dropna(subset=['zone_id']).groupby('zone_id').first().reset_index()
    stations = stations[['zone_id', 'stop_name', 'stop_lat', 'stop_lon']]

stations = stations.rename(columns={'zone_id': 'station_code'})
print(f"Number of stations: {len(stations)}")
print(stations.head(10))
```

```python
# Save stations
stations.to_csv('../data/processed/stations.csv', index=False)

# Join station coordinates to ridership data for both origin and destination
df = pd.read_csv('../data/processed/ridership_cleaned.csv')

# Join origin station coordinates
df = df.merge(stations, left_on='Origin Station', right_on='station_code', how='left')
df = df.rename(columns={'stop_name': 'origin_name', 'stop_lat': 'origin_lat', 'stop_lon': 'origin_lon'})
df = df.drop(columns=['station_code'])

# Join destination station coordinates
df = df.merge(stations, left_on='Destination Station', right_on='station_code', how='left')
df = df.rename(columns={'stop_name': 'dest_name', 'stop_lat': 'dest_lat', 'stop_lon': 'dest_lon'})
df = df.drop(columns=['station_code'])

print("Shape after join:", df.shape)
print(df.head())
print("\nNull counts:\n", df.isnull().sum())
```

```python
import pandas as pd
import glob, os

cols = ['Date', 'Hour', 'Origin Station', 'Destination Station', 'Trip Count']

kaggle_files = glob.glob('../data/raw/ridership/kaggle/*.csv')
bart_files = glob.glob('../data/raw/ridership/bart/*.csv')

all_files = kaggle_files + bart_files
first = True

for f in all_files:
    print(f"Processing {f}...")
    df = pd.read_csv(f, header=0, names=cols)
    df['Date'] = pd.to_datetime(df['Date'])
    df = df.dropna()
    df = df[df['Origin Station'] != df['Destination Station']]
    df['Year'] = df['Date'].dt.year
    df['Month'] = df['Date'].dt.month
    df['DayOfWeek'] = df['Date'].dt.dayofweek
    df['IsWeekend'] = df['DayOfWeek'].isin([5, 6]).astype(int)
    
    # Append to CSV one file at a time
    df.to_csv('../data/processed/ridership_cleaned.csv', 
              mode='w' if first else 'a',
              header=first,
              index=False)
    first = False
    print(f"  Done, shape: {df.shape}")

print("All files saved")
```

```python
import pandas as pd
import numpy as np

def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    return 2 * R * np.arcsin(np.sqrt(a))

chunks = []
chunk_size = 1_000_000

for chunk in pd.read_csv('../data/processed/ridership_with_stations.csv', chunksize=chunk_size):
    chunk['Date'] = pd.to_datetime(chunk['Date'])
    chunk['Week'] = chunk['Date'].dt.isocalendar().week.astype(int)
    
    # Calculate trip distance for each row
    chunk['trip_distance_km'] = haversine(
        chunk['origin_lat'], chunk['origin_lon'],
        chunk['dest_lat'], chunk['dest_lon']
    )
    
    # Aggregate to station-week level
    agg = chunk.groupby(['Year', 'Week', 'Month', 'Origin Station',
                         'origin_name', 'origin_lat', 'origin_lon']).agg(
        weekly_exits=('Trip Count', 'sum'),
        avg_trip_distance_km=('trip_distance_km', 'mean'),
        num_unique_destinations=('Destination Station', 'nunique'),
        is_weekend_ratio=('IsWeekend', 'mean')
    ).reset_index()
    
    chunks.append(agg)
    print(f"Processed chunk, shape: {agg.shape}")

print("Combining chunks...")
df_weekly = pd.concat(chunks, ignore_index=True)

# Re-aggregate since same station-week spans multiple chunks
df_weekly = df_weekly.groupby(['Year', 'Week', 'Month', 'Origin Station',
                                'origin_name', 'origin_lat', 'origin_lon']).agg(
    weekly_exits=('weekly_exits', 'sum'),
    avg_trip_distance_km=('avg_trip_distance_km', 'mean'),
    num_unique_destinations=('num_unique_destinations', 'sum'),
    is_weekend_ratio=('is_weekend_ratio', 'mean')
).reset_index()

print("Final shape:", df_weekly.shape)
print(df_weekly.head())
```

```python
# Fix num_unique_destinations - recompute from raw data
print("Recomputing unique destinations...")

dest_counts = []
for chunk in pd.read_csv('../data/processed/ridership_with_stations.csv', chunksize=1_000_000):
    chunk['Date'] = pd.to_datetime(chunk['Date'])
    chunk['Week'] = chunk['Date'].dt.isocalendar().week.astype(int)
    dest_counts.append(chunk.groupby(['Year', 'Week', 'Origin Station'])['Destination Station'].apply(set))

# Combine sets across chunks
print("Combining...")
dest_df = pd.concat(dest_counts).groupby(level=[0,1,2]).apply(lambda x: len(set().union(*x))).reset_index()
dest_df.columns = ['Year', 'Week', 'Origin Station', 'num_unique_destinations']

# Replace the wrong column in df_weekly
df_weekly = df_weekly.drop(columns=['num_unique_destinations'])
df_weekly = df_weekly.merge(dest_df, on=['Year', 'Week', 'Origin Station'], how='left')

print("Final shape:", df_weekly.shape)
print("Max unique destinations:", df_weekly['num_unique_destinations'].max())
print(df_weekly.head())
```

```python
df_weekly.to_csv('../data/processed/ridership_weekly.csv', index=False)
print("Saved successfully")
print(f"File shape: {df_weekly.shape}")
print(f"\nYears present: {sorted(df_weekly['Year'].unique())}")
print(f"\nWeekly exits range: {df_weekly['weekly_exits'].min()} to {df_weekly['weekly_exits'].max()}")
print(f"\nNull counts:\n{df_weekly.isnull().sum()}")
```

```python
import requests
import pandas as pd
#pull variable data from api
vars_str = ','.join(CENSUS_VARIABLES)

url = (
    f'https://api.census.gov/data/2023/acs/acs5'
    f'?get=NAME,{vars_str}'
    f'&for=tract:*'
    f'&in=state:06'
    f'&key={CENSUS_KEY}'
)

print("Pulling Census data for California tracts...")
response = requests.get(url)
data = response.json()

ca_tracts = pd.DataFrame(data[1:], columns=data[0])
print("Shape:", ca_tracts.shape)
print(ca_tracts.head())
```

```python
import geopandas as gpd
from shapely.geometry import Point

# Convert numeric columns
for col in CENSUS_VARIABLES:
    ca_tracts[col] = pd.to_numeric(ca_tracts[col], errors='coerce')

# Create a unique tract ID for joining
ca_tracts['tract_id'] = ca_tracts['state'] + ca_tracts['county'] + ca_tracts['tract']
print("Tracts shape:", ca_tracts.shape)

# Load stations
stations = pd.read_csv('../data/processed/stations.csv')

# Convert stations to GeoDataFrame
gdf_stations = gpd.GeoDataFrame(
    stations,
    geometry=[Point(lon, lat) for lon, lat in zip(stations['stop_lon'], stations['stop_lat'])],
    crs='EPSG:4326'
)

# Download California tract boundaries from Census TIGER
print("Downloading tract boundaries...")
tract_url = 'https://www2.census.gov/geo/tiger/TIGER2023/TRACT/tl_2023_06_tract.zip'
ca_tract_geo = gpd.read_file(tract_url)
print("Tract geometry shape:", ca_tract_geo.shape)
print(ca_tract_geo.head())

BAY_AREA_COUNTIES = [
    '001',  # Alameda
    '013',  # Contra Costa
    '041',  # Marin
    '055',  # Napa
    '075',  # San Francisco
    '081',  # San Mateo
    '085',  # Santa Clara
    '095',  # Solano
    '097',  # Sonoma
]

# Filter tract geometry to Bay Area only
ca_tract_geo = ca_tract_geo[ca_tract_geo['COUNTYFP'].isin(BAY_AREA_COUNTIES)]
print("Bay Area tracts shape:", ca_tract_geo.shape)

# Filter Census demographics to Bay Area only
ca_tracts = ca_tracts[ca_tracts['county'].isin(BAY_AREA_COUNTIES)]
print("Bay Area Census data shape:", ca_tracts.shape)
```

```python
BAY_AREA_COUNTIES = [
    '001',  # Alameda
    '013',  # Contra Costa
    '041',  # Marin
    '055',  # Napa
    '075',  # San Francisco
    '081',  # San Mateo
    '085',  # Santa Clara
    '095',  # Solano
    '097',  # Sonoma
]

# Filter tract geometry to Bay Area only
ca_tract_geo = ca_tract_geo[ca_tract_geo['COUNTYFP'].isin(BAY_AREA_COUNTIES)]
print("Bay Area tracts shape:", ca_tract_geo.shape)

# Filter Census demographics to Bay Area only
ca_tracts = ca_tracts[ca_tracts['county'].isin(BAY_AREA_COUNTIES)]
print("Bay Area Census data shape:", ca_tracts.shape)
print(ca_tracts.head())
print(ca_tract_geo.head())
```

```python
# Make sure both are in the same coordinate system
ca_tract_geo = ca_tract_geo.to_crs('EPSG:4326')

# Spatial join - find which tract each station falls inside
stations_with_tracts = gpd.sjoin(gdf_stations, ca_tract_geo[['GEOID', 'ALAND', 'geometry']], 
                                  how='left', predicate='within')

print("Shape:", stations_with_tracts.shape)
print("Null GEOIDs:", stations_with_tracts['GEOID'].isnull().sum())
print(stations_with_tracts[['station_code', 'stop_name', 'GEOID', 'ALAND']].head(10))
```

```python
# Create matching GEOID in ca_tracts
ca_tracts['GEOID'] = ca_tracts['state'] + ca_tracts['county'] + ca_tracts['tract']

# Join demographics to stations
stations_census = stations_with_tracts.merge(ca_tracts, on='GEOID', how='left')

# Convert all Census variables to numeric
for col in CENSUS_VARIABLES:
    stations_census[col] = pd.to_numeric(stations_census[col], errors='coerce')

# Engineer features
stations_census['population_density'] = stations_census['B01003_001E'] / (stations_census['ALAND'] / 1_000_000)
stations_census['low_car_rate'] = (stations_census['B08201_002E'] + stations_census['B08201_003E']) / stations_census['B08201_001E']
stations_census['transit_commuter_rate'] = stations_census['B08301_010E'] / stations_census['B08301_001E']
stations_census['driver_rate'] = stations_census['B08301_002E'] / stations_census['B08301_001E']
stations_census['carpool_rate'] = stations_census['B08301_004E'] / stations_census['B08301_001E']
stations_census['employment_rate'] = stations_census['B23025_004E'] / stations_census['B23025_001E']
stations_census['renter_rate'] = stations_census['B25003_003E'] / stations_census['B25003_001E']
stations_census['median_income'] = stations_census['B19013_001E']

print("Shape:", stations_census.shape)
print("\nEngineered features:")
print(stations_census[['station_code', 'stop_name', 'population_density', 'low_car_rate', 
                         'transit_commuter_rate', 'employment_rate', 'renter_rate', 
                         'median_income']].to_string())
```

```python
# Deal with SFO outlier
# Replace Census placeholder values
import numpy as np
stations_census = stations_census.replace(-666666666, np.nan)

# Check nulls
print(stations_census[['station_code', 'population_density', 'low_car_rate', 
                         'transit_commuter_rate', 'renter_rate', 'median_income']].isnull().sum())

stations_census = stations_census[stations_census['station_code'] != 'SFIA']
print("Shape after dropping SFO:", stations_census.shape)
print("Remaining nulls:", stations_census[['low_car_rate', 'transit_commuter_rate', 
                                            'renter_rate', 'median_income']].isnull().sum().sum())
```
